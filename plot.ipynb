```python
import uproot
import awkward as ak
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

# Open the ROOT files
file1 = uproot.open("sm_delphes.root")
file2 = uproot.open("heft_Higgs.root")

# Extract Jet.PT from both trees
tree1 = file1["Delphes"]
tree2 = file2["Delphes"]
jet_pt1 = tree1['Jet.PT'].array()
jet_pt2 = tree2['Jet.PT'].array()

# Flatten the arrays
jet_pt1_flat = ak.flatten(jet_pt1)
jet_pt2_flat = ak.flatten(jet_pt2)

# Plot Jet PT from both ROOT files with normalization and Gaussian fit
plt.figure(figsize=(12, 6))
plt.yscale('log')

# Normalize the histograms
counts1, bins1, _ = plt.hist(jet_pt1_flat, bins=50, density=True, alpha=0.5, label='SM')
counts2, bins2, _ = plt.hist(jet_pt2_flat, bins=50, density=True, alpha=0.5, label='SMEFT')

# Fit Gaussian curves
mu1, std1 = norm.fit(jet_pt1_flat)
mu2, std2 = norm.fit(jet_pt2_flat)

# Plot the Gaussian curves
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p1 = norm.pdf(x, mu1, std1)
p2 = norm.pdf(x, mu2, std2)
plt.plot(x, p1, 'k', linewidth=2, label=f'Gaussian Fit SM: $\mu={mu1:.2f}$, $\sigma={std1:.2f}$')
plt.plot(x, p2, 'r', linewidth=2, label=f'Gaussian Fit SMEFT: $\mu={mu2:.2f}$, $\sigma={std2:.2f}$')

plt.xlabel('Jet PT')
plt.ylabel('Frequency')
plt.title('Jet PT Distribution with Gaussian Fit')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
jet_eta1 = tree1['Jet.Eta'].array()
jet_eta2 = tree2['Jet.Eta'].array()

# Flatten the arrays if not already flattened
jet_eta1_flat = ak.flatten(jet_eta1)
jet_eta2_flat = ak.flatten(jet_eta2)

# Plot Jet Eta from both ROOT files with normalization and Gaussian fit
plt.figure(figsize=(12, 6))
# plt.yscale('log')

# Normalize the histograms
counts_eta1, bins_eta1, _ = plt.hist(jet_eta1_flat, bins=50, density=True, alpha=0.5, label='SM')
counts_eta2, bins_eta2, _ = plt.hist(jet_eta2_flat, bins=50, density=True, alpha=0.5, label='SMEFT')

# Fit Gaussian curves
mu_eta1, std_eta1 = norm.fit(jet_eta1_flat)
mu_eta2, std_eta2 = norm.fit(jet_eta2_flat)

# Plot the Gaussian curves
xmin_eta, xmax_eta = plt.xlim()
x_eta = np.linspace(xmin_eta, xmax_eta, 100)
p_eta1 = norm.pdf(x_eta, mu_eta1, std_eta1)
p_eta2 = norm.pdf(x_eta, mu_eta2, std_eta2)
plt.plot(x_eta, p_eta1, 'k', linewidth=2, label=f'Gaussian Fit SM: $\mu={mu_eta1:.2f}$, $\sigma={std_eta1:.2f}$')
plt.plot(x_eta, p_eta2, 'r', linewidth=2, label=f'Gaussian Fit SMEFT: $\mu={mu_eta2:.2f}$, $\sigma={std_eta2:.2f}$')

plt.xlabel('Jet Eta')
plt.ylabel('Frequency')
plt.title('Jet Eta Distribution with Gaussian Fit')
plt.legend()

plt.tight_layout()
plt.show()
```
